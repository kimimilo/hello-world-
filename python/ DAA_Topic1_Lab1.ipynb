```python
# for loops go through a sequence a known number of times (each item in a list, each number in a range).
# while loops keep going as long as a condition stays true.
# continue skips to the next round; break exits the loop completely.

urls = ["site.com/page1", "site.com/page2", "BROKEN", "site.com/page4", "BLOCKED", "site.com/page6"]

for url in urls:
    if url == "BROKEN":
        print("Skipping a broken link")
        continue
    elif url == "BLOCKED":
        print("Server blocked us, stopping the scraper")
        break
    else:
        print("Scraping", url)

# while loop: keep doubling savings until they pass a target
savings, months = 100, 0
while savings < 1000:
    savings = savings * 2
    months = months + 1
print(f"Reached {savings} after {months} months")
```

```python
# Lists
# Python uses zero-based indexing: the first item is at position 0. Slicing my_list[start:end] includes start but stops before end
# [] used when making lists, indexing - accesing single elements of a sequence , slicing - extracing single parts of a sequence
prices = [10, 20, 30, 40, 50]
print(prices[1])        # indexing -> 20
print(prices[1:3])      # slicing  -> [20, 30]
print(prices[-1])       # negative index counts from the end -> 50

prices.append(60)       # add to the end
prices.remove(10)       # remove the first matching value
print(prices, "has", len(prices), "items") # len(prices) counts the number of items in a list
```

```python
# """" allows the code to spam multiple lines
raw_csv = """product,units,price
Notebook,12,3.50
Pen,40,1.20
Backpack,3,45.00
Calculator,5,22.90"""

# .split used to split a stirng into a list of smaller strings

rows = raw_csv.split("\n")  # split("\n") then means to cut the string into new lines whenever
header = rows[0].split(",") # header is like the name on top, products, units, price
                             # rows [0] gets the first item in the list as pyhton starts from 0, which is products price unit
                             # the (",") cuts whenevers there's a comma --> ["product", "units", "price"] this is what we get

print("columns", header)     # looks like columns ['product', 'units', 'price'] --> then it continues downwards

total_revenue = 0 # starts the revenue total at 0
prices = [] # creates an empty list to store the product prices into
for row in rows[1:]: # means get everything from index 1 to the end, dont need start from 0 this time as dont need calculate the words product...
                     # repeats the loop as it goes through each row , used when looking through multiple rows of data

# all the way until the last 2 lines here are indented 4 spaces because they're part of the loop you want to be repeated
    product, units, price = row.split(",") # DIFF from 1st .split, this is sued to extract the values to sort out vs extracting the names as a header units = int(units) # to convert unit & price into intergers and floats as they are still strings due to the "", necessary for multiplication
    product, units, price = row.split(",")
    units = int(units)
    price = float(price)
    revenue = units * price

    prices.append(price)    # add an item to the very end of an existing list
    total_revenue += revenue # += ia a shorter way of writing total_revenue = total_revenue + revenue
    print(f"{product:<11} {units:>3} x {price:> 6.2f} = {revenue:>7.2f}") # f lets python easily insert variables into the text
                                                                          # numbers here are based on how long the words are, usually + 1 just for extra space
                                                                          # decimal points are INCLUDED when counting the number of space taken up
                                                                          # DIFFERENT from >= , without the equal sign it means to align
                                                                          # .2f means to decimal points on top of 6 characters to the right


print(f"total revenue: {total_revenue:.2f}")
print(f"average price: {sum(prices) / len(prices):.2f}") # basically sum / number of items in the list
```

```python
# Extract all scores. Calculate the average score. Count how many students passed (score ≥ 70). Print a formatted summary.

students = [
    {"name": "Aisha", "score": 88},
    {"name": "Ben", "score": 64},
    {"name": "Chen", "score": 91},
    {"name": "Divya", "score": 70},
    {"name": "Ethan", "score": 55},
]

scores = [s["score"] for s in students]      # a list comprehension: build a list in one line
average = sum(scores) / len(scores)

passed = 0
for s in students:
    if s["score"] >= 70:
        passed += 1

print(f"Students: {len(students)}")
print(f"Average score: {average:.1f}")
print(f"Passed: {passed} | Failed: {len(students) - passed}")
print("Top scorer:", max(students, key=lambda s: s["score"])["name"])
```
